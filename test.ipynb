```python
import matplotlib.pyplot as plt

# Dữ liệu các năm (loại bỏ 2022 theo yêu cầu)
years = ['2017', '2018', '2019', '2020', '2021', '2023', '2024', '2025']

# Dữ liệu Doanh thu & Lợi nhuận (Đơn vị: Tỷ VNĐ)
revenue = [611.8, 639.1, 691.4, 728.5, 650.1, 719.5, 727.0, 777.8]
profit = [36.4, 38.2, 46.9, 42.1, 34.6, 39.6, 39.2, 50.8]

# Khởi tạo đồ thị với 2 khung (2 biểu đồ trên dưới)
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 12))

# ----------------------------------------------------
# 1. BIỂU ĐỒ DOANH THU THUẦN
# ----------------------------------------------------
bars1 = ax1.bar(years, revenue, color='#4C72B0', edgecolor='black')
ax1.set_title('XU HƯỚNG DOANH THU THUẦN (TỶ VNĐ)\n(Giai đoạn 2017 - 2025 (trừ 2022))', fontsize=14, fontweight='bold', pad=15)
ax1.set_ylabel('Tỷ VNĐ', fontsize=12)
ax1.set_ylim(0, 900) # Thiết lập giới hạn trục Y để biểu đồ có không gian thở
ax1.grid(axis='y', linestyle='--', alpha=0.7)

# Ghi chú giá trị trên đầu mỗi cột doanh thu
for bar in bars1:
    yval = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2, yval + 15, f'{yval:.1f}', ha='center', va='bottom', fontsize=11, fontweight='bold')

# ----------------------------------------------------
# 2. BIỂU ĐỒ LỢI NHUẬN SAU THUẾ
# ----------------------------------------------------
bars2 = ax2.bar(years, profit, color='#55A868', edgecolor='black')
ax2.set_title('QUỸ ĐẠO LỢI NHUẬN SAU THUẾ (TỶ VNĐ)\n(Giai đoạn 2017 - 2025 (trừ 2022))', fontsize=14, fontweight='bold', pad=15)
ax2.set_ylabel('Tỷ VNĐ', fontsize=12)
ax2.set_ylim(0, 60) # Thiết lập giới hạn trục Y
ax2.grid(axis='y', linestyle='--', alpha=0.7)

# Ghi chú giá trị trên đầu mỗi cột lợi nhuận
for bar in bars2:
    yval = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2, yval + 1, f'{yval:.1f}', ha='center', va='bottom', fontsize=11, fontweight='bold')

# ----------------------------------------------------
# 3. HOÀN THIỆN VÀ LƯU ẢNH
# ----------------------------------------------------
plt.tight_layout(pad=3.0)
plt.savefig('bieu_do_taichinh_A32.png', dpi=300, bbox_inches='tight')
```
