Dọn dẹp và xử lý 

```python
import pandas as pd
import re

# 1. Đọc trực tiếp file DATASET GỐC
raw_df = pd.read_csv('AirlineScrappedReview.csv')

# 2. Hàm Cứu hộ dữ liệu Tối thượng
def rescue_full_data(row):
    # Khởi tạo bộ khung giữ lại TẤT CẢ các cột quan trọng
    new_row = {
        'Passanger_Name': row['Passanger_Name'],
        'Flying_month': None,
        'Route': None,
        'Rating': row['Rating'],
        'Verified': 'Unknown',
        'Review_title': row['Review_title'],
        'Review_content': row['Review_content'],
        'Traveller_type': None,
        'Class': None,
        'Aircraft': None  # Cột mới tách máy bay
    }
    
    # --- XỬ LÝ LỖI "VERIFIED" BỊ DÍNH VÀO REVIEW_CONTENT TỪ DÒNG 1273 ---
    content = str(row['Review_content'])
    if 'Verified Review' in content:
        new_row['Verified'] = 'Trip Verified'
        # Cắt bỏ chữ Verified ra khỏi nội dung bình luận
        new_row['Review_content'] = content.replace('Verified Review', '').replace('|', '').strip()
    elif 'Not Verified' in content or 'Unverified' in content:
        new_row['Verified'] = 'Not Verified'
        new_row['Review_content'] = content.replace('Not Verified', '').replace('Unverified', '').replace('|', '').strip()
    # Nếu cột Verified cũ có chứa chữ thì vẫn lấy
    elif pd.notna(row['Verified']) and ('Verified' in str(row['Verified'])):
        new_row['Verified'] = str(row['Verified'])
        
    # --- GỠ RỐI CÁC CỘT BỊ XÔ LỆCH ---
    # Chỉ quét các cột dễ bị lệch để nhặt dữ liệu
    search_cols = ['Flying_month', 'Route', 'Traveller_type', 'Class']
    vals = [str(row[c]).strip() for c in search_cols if pd.notna(row[c])]
    
    month_regex = re.compile(r'^[A-Z][a-z]+\s\d{4}$')
    
    for val in vals:
        if val == 'nan': continue
        
        if month_regex.match(val):
            new_row['Flying_month'] = val
        elif 'Class' in val:
            new_row['Class'] = val
        elif 'Leisure' in val or val == 'Business':
            new_row['Traveller_type'] = val
        # Nhận diện Máy bay để không bị lẫn vào Traveller_type
        elif any(kw in val for kw in ['Boeing', 'A3', 'Embraer', '737', '747', '777', '787', 'neo', 'MAX']):
            new_row['Aircraft'] = val
        elif ' to ' in val and len(val) < 50:
            new_row['Route'] = val
            
    return pd.Series(new_row)

# 3. Áp dụng thuật toán lên toàn bộ data
clean_full_df = raw_df.apply(rescue_full_data, axis=1)

# 4. Xuất ra file chuẩn hóa cuối cùng
clean_full_df.to_csv('Cleaned_Airline_Review_Full.csv', index=False)

print(f"Hoàn tất! Tổng số mẫu được giữ lại: {len(clean_full_df)} dòng (Không rớt 1 dòng nào).")
print("Đã giữ nguyên 3 cột Passanger_Name, Review_title, Review_content.")
print("Đã tách thành công Aircraft và sửa lỗi Verified.")
```
